```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Load your local ABS data
df_abs = pd.read_csv('Cleaned_Banking_Data.csv')
# Standardise ABS dates to the start of the month
df_abs['Date'] = pd.to_datetime(df_abs['Date'], dayfirst=True).dt.to_period('M').dt.to_timestamp()

# 2. Use the CORRECT direct CSV link for RBA Cash Rate (Table F1.1)
rba_csv_url = "https://rba.gov.au"
# We skip 10 rows because RBA files have a long text header
df_rba = pd.read_csv(rba_csv_url, skiprows=9)

# 3. Clean RBA columns and align dates
df_rba.rename(columns={df_rba.columns[0]: 'Date', df_rba.columns[1]: 'Cash_Rate'}, inplace=True)
df_rba['Date'] = pd.to_datetime(df_rba['Date'], dayfirst=True).dt.to_period('M').dt.to_timestamp()

# 4. Average the daily RBA rates into monthly rates to match ABS
df_rba_monthly = df_rba.groupby('Date')['Cash_Rate'].mean().reset_index()

# 5. Merge the two tables
df = pd.merge(df_abs, df_rba_monthly, on='Date', how='inner')

# 6. Run the Predictive Model
# Using your series ID from the previous error printout
y_col = 'A130268677T' 
df_ml = df.dropna(subset=['Cash_Rate', y_col])

X = df_ml[['Cash_Rate']]
y = df_ml[y_col]

model = LinearRegression().fit(X, y)

print(f"✅ Success! Data merged using {len(df_ml)} monthly records.")
print(f"Model Accuracy (R^2): {model.score(X, y):.2f}")

# Predict for a 5% Cash Rate
pred = model.predict([[5.0]])
print(f"At a 5% Cash Rate, predicted lending ({y_col}) is: ${pred[0]:,.2f}")
```

```python
import pandas as pd

# --- STEP 1: LOAD RAW DATA ---
# (Ensure these 3 files are in your folder)
t1 = pd.read_excel('560101.xlsx', sheet_name='Data1', skiprows=9)
t2 = pd.read_excel('560102.xlsx', sheet_name='Data1', skiprows=9)
t32 = pd.read_excel('560132.xlsx', sheet_name='Data1', skiprows=9)

# --- STEP 2: STANDARDIZE COLUMNS ---
for df in [t1, t2, t32]:
    # Name the first column 'Date'
    df.rename(columns={df.columns[0]: 'Date'}, inplace=True)
    # Convert to standard Python date objects
    df['Date'] = pd.to_datetime(df['Date'])

# --- STEP 3: CREATE PATH B (HISTORICAL: 2002-2024) ---
# Merge Owner-Occupier (T1) and Investor (T2)
df_historical = pd.merge(t1, t2, on='Date', how='inner')
print(f"Path B (Historical) Rows: {len(df_historical)}")

# --- STEP 4: CREATE PATH A (MODERN: 2019-2024) ---
# Merge the above with Business (T32). 
# 'inner' join clips the start date to July 2019 automatically.
df_modern = pd.merge(df_historical, t32, on='Date', how='inner')
print(f"Path A (Modern) Rows: {len(df_modern)}")

# --- STEP 5: VERIFY ---
df_modern.head()
```

```python
import pandas as pd

# 1. Load the RBA Cash Rate (CSV)
rba_df = pd.read_csv('f1.1-data.csv', skiprows=10)
rba_df.rename(columns={rba_df.columns[0]: 'Date', rba_df.columns[1]: 'Cash_Rate'}, inplace=True)
rba_df['Date'] = pd.to_datetime(rba_df['Date'])

# 2. THE MASTER MERGE (Historical: 2002 - Sept 2024)
# We join your already merged df_historical with the RBA rates
df_hist_final = pd.merge(df_historical, rba_df[['Date', 'Cash_Rate']], on='Date', how='inner')

# 3. THE MASTER MERGE (Modern: 2019 - Sept 2024)
# We join your already merged df_modern with the RBA rates
df_mod_final = pd.merge(df_modern, rba_df[['Date', 'Cash_Rate']], on='Date', how='inner')

# 4. THE "CLEAN CUT" (Filtering to Sept 2024)
df_hist_final = df_hist_final[df_hist_final['Date'] <= '2024-09-01']
df_mod_final = df_mod_final[df_mod_final['Date'] <= '2024-09-01']

# 5. EXPORT TO PHYSICAL FILES
df_hist_final.to_csv('Final_Historical_Data.csv', index=False)
df_mod_final.to_csv('Final_Modern_Data.csv', index=False)

print(f"Data Engineering Complete!")
print(f"Historical Dataset: {df_hist_final.shape} rows (ending Sept 2024)")
print(f"Modern Dataset: {df_mod_final.shape} rows (ending Sept 2024)")
```

```python
import pandas as pd

# 1. Load RBA with the 'dayfirst' fix
rba_df = pd.read_csv('f1.1-data.csv', skiprows=10)
rba_df.rename(columns={rba_df.columns[0]: 'Date', rba_df.columns[1]: 'Cash_Rate'}, inplace=True)

# FIX A: Tell Python it is Day-Month-Year
rba_df['Date'] = pd.to_datetime(rba_df['Date'], dayfirst=True)

# FIX B: Force RBA dates to the 1st of the month so they match ABS (e.g., 30-06 becomes 01-06)
rba_df['Date'] = rba_df['Date'].dt.to_period('M').dt.to_timestamp()

# 2. Re-run the Master Merge
# (Using inner join now that dates are aligned)
df_hist_final = pd.merge(df_historical, rba_df[['Date', 'Cash_Rate']], on='Date', how='inner')
df_mod_final = pd.merge(df_modern, rba_df[['Date', 'Cash_Rate']], on='Date', how='inner')

# 3. Apply the 'Sept 2024' Cut-off
df_hist_final = df_hist_final[df_hist_final['Date'] <= '2024-09-01']
df_mod_final = df_mod_final[df_mod_final['Date'] <= '2024-09-01']

print(f"Data Engineering Complete!")
print(f"Historical Dataset: {df_hist_final.shape} rows")
print(f"Modern Dataset: {df_mod_final.shape} rows")
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# 1. Prepare the 'X' (Cause) and 'y' (Effect)
# We use the Historical data because it has the strongest correlation
X = df_hist_final[['Cash_Rate']].values
y = df_hist_final[target_id].values

# 2. Create and Train the Model
model = LinearRegression()
model.fit(X, y)

# 3. Get the "Sensitivity" (The Coefficient)
# This tells us exactly how many Millions ($m) lending drops for every 1% rate hike
sensitivity = model.coef_[0]

print(f"--- MACHINE LEARNING MODEL READY ---")
print(f"Impact: For every 1% increase in Cash Rate, Lending drops by: ${abs(sensitivity):.2f} Million")
```

```python
import matplotlib.pyplot as plt

# 1. Create a "Line of Best Fit" using your ML model
y_pred = model.predict(X)

# 2. Plot the Actual Data (Blue Dots) vs. The ML Prediction (Red Line)
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', alpha=0.3, label='Actual Monthly Data')
plt.plot(X, y_pred, color='red', linewidth=3, label='Machine Learning Prediction')

# 3. Label it for a Professional Look
plt.title('Mortgage Demand Sensitivity to RBA Cash Rate (2002-2024)')
plt.xlabel('RBA Cash Rate (%)')
plt.ylabel('Monthly Lending Volume ($ millions)')
plt.legend()
plt.grid(True)
plt.show()
```
